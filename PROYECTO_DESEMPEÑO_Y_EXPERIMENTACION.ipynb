## **Monitores a utilizar**

### Linux (Ubuntu 24.04)

#### MangoHud (FPS y Frametime)

MangoHud es una herramienta de código abierto diseñada para el monitoreo de videojuegos en tiempo real. Permite visualizar y registrar métricas relacionadas con el desempeño gráfico de una aplicación, incluyendo los fotogramas por segundo (FPS) y los tiempos de fotograma (frametime).

Para este experimento, MangoHud será utilizado para medir las variables de respuesta relacionadas con el desempeño del juego Don't Starve Together, específicamente FPS y frametime.

#### Psutil (RAM, Swap, CPU y Disco)

Psutil es una biblioteca multiplataforma para Python que permite obtener información detallada sobre procesos y recursos del sistema operativo. La herramienta proporciona acceso a métricas relacionadas con el uso de CPU, memoria RAM, memoria virtual (swap) y actividad de disco.

Durante las pruebas en Ubuntu, Psutil será utilizado para registrar el uso de memoria RAM del proceso del juego, el uso de memoria RAM del sistema, el uso de memoria swap y la actividad de entrada/salida del disco. Estas métricas serán almacenadas automáticamente para su posterior análisis.

### Windows 11

#### MSI Afterburner + RivaTuner Statistics Server (FPS y Frametime)

MSI Afterburner es una herramienta de monitoreo de hardware ampliamente utilizada para la supervisión del rendimiento de videojuegos. En conjunto con RivaTuner Statistics Server (RTSS), permite visualizar y registrar métricas de desempeño en tiempo real.

Para este experimento, estas herramientas serán utilizadas para capturar los FPS y frametimes generados durante la ejecución de Don't Starve Together.

#### Psutil (RAM, Swap, CPU y Disco)

Psutil también será utilizado en Windows 11 con el objetivo de mantener un método de medición consistente entre ambos sistemas operativos. La biblioteca permitirá registrar el uso de memoria RAM del proceso monitoreado, el uso de memoria RAM del sistema, la utilización de memoria virtual (pagefile/swap), el uso de CPU y la actividad de disco durante cada tratamiento experimental.

La utilización de una misma biblioteca en ambos sistemas operativos permite reducir diferencias metodológicas entre tratamientos y facilita la automatización de la captura de datos mediante una única base de código desarrollada en Python.

## **Diseño final del experimento**

Para este último avance me dediqué principalmente a consolidar y automatizar las herramientas utilizadas durante la recolección y el procesamiento de los datos experimentales.

Inicialmente, ya contaba con las herramientas necesarias para obtener las mediciones en Linux. Estas consistían en el script **metricasPsutil.py**, encargado de registrar métricas del sistema, y en **mergeLinux.py** (actualmente renombrado como **mergeMetricas.py**), cuya función era combinar los archivos generados por **metricasPsutil.py** con los archivos producidos por MangoHud, que contienen las métricas de FPS y *frametime*. La unión de ambos conjuntos de datos se realizaba utilizando la columna **elapsed** como referencia temporal.

En su versión original, **mergeLinux.py** requería que el usuario indicara manualmente cada par de archivos y la ruta donde se almacenaría el resultado, lo que hacía que el proceso fuera lento y propenso a errores cuando se trabajaba con un gran número de mediciones. Por esta razón, el script fue modificado para que procesara automáticamente todos los archivos contenidos en una carpeta. Para ello, primero identifica cuáles corresponden a **metricasPsutil.py** y cuáles a MangoHud; posteriormente, utilizando la fecha y hora de inicio del monitoreo presentes en el nombre de los archivos, empareja automáticamente aquellos cuya diferencia temporal es mínima. Finalmente, aplica una corrección sobre la columna **elapsed** para compensar el desfase existente entre ambas herramientas y realizar la combinación de los datos de forma adecuada.

Posteriormente, el script también fue adaptado para permitir la integración de mediciones provenientes de Windows, utilizando los registros generados por MSI Afterburner. Como parte de este proceso, se desarrolló el programa **MSIAfterBurnerToCSV.py**, el cual convierte los archivos HML generadores por MSI Afterburner al formato CSV y, además, automatiza parte del proceso de recolección de métricas durante las pruebas.

Durante la integración de los datos obtenidos en Windows surgieron algunos inconvenientes relacionados con diferencias de formato respecto a los archivos generados en Linux. Para solucionarlos se desarrolló el script **correcciones.py**, encargado de corregir inconsistencias detectadas, como el uso de la columna **elapse** en lugar de **elapsed** y diferencias en el formato de la fecha presente en los nombres de los archivos.

Uno de los principales desafíos del experimento fue la limitación de memoria en Windows. A diferencia de Linux, donde es posible utilizar *cgroups* para restringir la memoria disponible para un proceso y sus procesos hijos, Windows no ofrece un mecanismo equivalente de uso tan directo. Debido al tiempo disponible para el proyecto y a que no contaba con experiencia previa en el lenguaje C# ni con la API de bajo nivel *Job Objects*, recurrí al apoyo de ChatGPT para desarrollar un script que permitiera establecer dichos límites de memoria. Esta herramienta resultó de gran utilidad, ya que permitió implementar la funcionalidad requerida y continuar con las pruebas experimentales sin necesidad de desarrollar una solución desde cero.


```python
from pathlib import Path
import pandas as pd

resumen = []

tratamientos = {
    "Linux SL": "/content/Proyecto_de_monitoreo_DE/treatment_Linux/treatment_SL/mergeSL",
    "Linux 3600": "/content/Proyecto_de_monitoreo_DE/treatment_Linux/treatment_3600/merge3600",
    "Linux 3800": "/content/Proyecto_de_monitoreo_DE/treatment_Linux/treatment_3800/merge3800",

    "Windows SL": "/content/Proyecto_de_monitoreo_DE/treatment_Windows/treatment_SL/mergeSL",
    "Windows 3600": "/content/Proyecto_de_monitoreo_DE/treatment_Windows/treatment_3600/merge3600",
    "Windows 3800": "/content/Proyecto_de_monitoreo_DE/treatment_Windows/treatment_3800/merge3800",
}

for tratamiento, carpeta in tratamientos.items():

    for archivo in Path(carpeta).glob("*_merge.csv"):

        df = pd.read_csv(archivo)

        resumen.append({
            "tratamiento": tratamiento,
            "fps": df["fps"].mean(),
            "RAM": df["RAM"].mean(),
            "CPU": df["CPU"].mean(),
            "frametime": df["frametime"].mean(),
            "SWAP": df["SWAP_OS"].mean(),
            "I/O_output": df["I/O_output"].mean(),
            "I/O_input": df["I/O_input"].mean(),
        })

df_resumen = pd.DataFrame(resumen)

df_resumen_comprimido = df_resumen.groupby("tratamiento").agg({
    "fps": "mean",
    "RAM": "mean",
    "CPU": "mean",
    "frametime": "mean",
    "SWAP": "mean"
}).reset_index()

display(df_resumen_comprimido)
```

```python
import matplotlib.pyplot as plt

df_resumen.boxplot(column="fps", by="tratamiento", figsize=(12,8))

plt.title("Distribución del FPS promedio por tratamiento")
plt.suptitle("")

plt.xlabel("Tratamiento")
plt.ylabel("FPS promedio")

plt.grid(True)

plt.show()
```

Estas primeras escaramuzas con los datos muestran lo que ya suponíamos inicialmente: en lo que respecta a mantener FPS altos, Ubuntu sigue mostrando un mejor desempeño que Windows 11 bajo las condiciones evaluadas. La diferencia observada llega incluso a alcanzar aproximadamente los 10 FPS entre ambos sistemas operativos.

Esto sugiere que Ubuntu logra manejar de mejor manera los recursos necesarios para la ejecución del juego, permitiendo mantener una mayor tasa de cuadros. Sin embargo, esta es una primera observación basada principalmente en FPS

```python
import seaborn as sns

plt.figure(figsize=(12,6))

sns.boxplot(
    data=df_resumen,
    x="tratamiento",
    y="RAM"
)

plt.xticks(rotation=45)
plt.title("Consumo promedio de RAM")
plt.ylabel("RAM utilizada")

plt.show()
```

Ahora bien, esto resulta interesante. En lo que respecta a Linux, parece que mientras más reducíamos la memoria RAM disponible, menor era el uso de memoria observado por el sistema. En cambio, Windows presentó un comportamiento contrario: mientras más reducíamos la memoria disponible, mayor parecía ser el uso de RAM por parte del proceso.

Este comportamiento resulta llamativo, ya que podría indicar diferencias en la forma en que ambos sistemas operativos administran la presión de memoria. Una posible explicación es que Windows intenta mantener más información del juego en memoria física mientras sea posible, evitando recurrir al archivo de paginación hasta que sea necesario.

También podría existir un comportamiento propio de Steam o del motor del juego, donde al detectar suficiente memoria disponible intenta mantener más recursos cargados para evitar accesos posteriores al disco. En este caso, el aumento del uso de RAM no necesariamente representaría un problema, sino una estrategia para mejorar tiempos de acceso y mantener estabilidad en la ejecución.

```python
import seaborn as sns

plt.figure(figsize=(12,6))

sns.boxplot(
    data=df_resumen,
    x="tratamiento",
    y="SWAP"
)

plt.xticks(rotation=45)
plt.title("Consumo promedio de SWAP")
plt.ylabel("SWAP utilizada")

plt.show()
```

Esta es la observación más llamativa hasta el momento. En el caso de Linux, cuando no existía ninguna limitación parecía utilizar una cantidad considerable de memoria virtual o swap en comparación con los tratamientos donde se fue reduciendo la memoria disponible. Sin embargo, al aplicar el límite de 3800 MB, el uso de memoria virtual disminuyó, mientras que al reducir nuevamente la memoria disponible a 3600 MB volvió a aumentar, llegando incluso a superar el tratamiento sin límites.

Este comportamiento resulta interesante debido a que no sigue un patrón completamente lineal, donde una menor cantidad de memoria disponible implique necesariamente un aumento constante del uso de swap. Parece existir un punto intermedio donde el comportamiento del sistema cambia y la administración de memoria toma una estrategia diferente.

Por otro lado, Windows presentó un comportamiento diferente. Tanto el tratamiento sin límites como el de 3600 MB mantuvieron un uso relativamente bajo de memoria virtual; sin embargo, el caso de 3800 MB resulta precisamente el más extraño, ya que siendo un tratamiento intermedio terminó utilizando más memoria virtual que cualquiera de los otros escenarios.

Estos resultados sugieren que la relación entre la memoria asignada, el uso de RAM y la utilización de memoria virtual depende de la estrategia de administración de memoria del sistema operativo. No necesariamente una mayor restricción provoca siempre mayor uso de swap, sino que puede existir un punto donde el sistema cambia la forma en que conserva, libera o pagina información.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))

ax = sns.scatterplot(
    data=df_resumen,
    x="SWAP",
    y="frametime",
    hue="tratamiento"
)

ax.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.title("Relación entre uso de Swap y Frametime")
plt.xlabel("Uso de Swap")
plt.ylabel("Frametime (ms)")

plt.show()
```

Este gráfico nos termina por mostrar una relación interesante: bajo las condiciones evaluadas, el uso de swap parece estar asociado con mejores valores de frametime. Un ejemplo claro se observa en Windows, donde parece existir una fuerte tendencia a mantener la mayor cantidad posible de información en RAM y reducir el uso de memoria virtual; sin embargo, este comportamiento parece haber tenido un costo reflejado en el frametime obtenido.

Esto resulta llamativo, ya que normalmente se espera que el uso de memoria virtual tenga un impacto negativo debido a la diferencia de velocidad entre la RAM y el almacenamiento secundario. No obstante, los resultados sugieren que una política más agresiva de mantener todo en memoria física no necesariamente garantiza un mejor rendimiento, ya que puede aumentar la presión sobre la RAM disponible.

Por lo tanto, en este experimento el uso de swap no debe interpretarse únicamente como una penalización, sino como una herramienta de administración de memoria que puede ayudar al sistema a distribuir mejor los recursos dependiendo del nivel de presión existente.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))

ax = sns.scatterplot(
    data=df_resumen,
    x="RAM",
    y="SWAP",
    hue="tratamiento"
)

ax.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.title("Relación entre uso de la RAM y SWAP ")
plt.xlabel("Uso de RAM")
plt.ylabel("Uso de SWAP")

plt.show()
```

Este gráfico evidencia una diferencia interesante en la forma en que Linux y Windows administran la memoria disponible. En el caso de Linux, se observa un comportamiento más flexible respecto al uso de memoria virtual, utilizando swap con mayor facilidad incluso cuando todavía existe memoria RAM disponible.

Esto sugiere un comportamiento adaptativo del sistema, donde Linux parece permitir una mayor utilización de swap como parte de su estrategia de administración de memoria, en lugar de intentar mantener la mayor cantidad posible de información exclusivamente en RAM.

Por otro lado, Windows parece presentar una tendencia más conservadora respecto al uso de memoria virtual, priorizando mantener la información en memoria física antes de recurrir al swap. Aunque esto puede reducir la dependencia del almacenamiento secundario, también podría generar mayor presión sobre la RAM disponible bajo escenarios limitados.
